# **DATA PRE-PROCESSING**

# **Data Quality Assessment**

### Missing Value Analysis

### Purpose

### Identify missing values in each column of the raw dataset before performing data cleaning.

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

## **Observation**

### The raw dataset contains missing values in five columns:

### - Rating – 1,474 missing values

# **Handling Missing Values**

### **Purpose**

### Handle the missing values identified during the data quality assessment using appropriate imputation methods.

```python
# Fill missing Rating values using the median
df["Rating"] = df["Rating"].fillna(df["Rating"].median())

# Fill missing categorical values using the mode
df["Type"] = df["Type"].fillna(df["Type"].mode()[0])
df["Content Rating"] = df["Content Rating"].fillna(df["Content Rating"].mode()[0])
df["Current Ver"] = df["Current Ver"].fillna(df["Current Ver"].mode()[0])
df["Android Ver"] = df["Android Ver"].fillna(df["Android Ver"].mode()[0])
```

```python
print("Missing values after handling:")
print(df.isnull().sum())
```

###  **Observation**

### The missing values in the Rating, Type, Content Rating, Current Ver, and Android Ver columns were successfully handled using appropriate imputation methods.

### The numerical Rating values were replaced using the median, while missing categorical values were replaced using their respective modes.

### After this step, these columns contain no missing values. The Size column will be examined separately during data-type conversion because values such as Varies with device are stored as text rather than as standard missing values.

# **Checking Duplicate Records**

### Purpose

### Identify duplicate records in the dataset before further data cleaning and preprocessing.

### Duplicate records can affect statistical analysis and may lead to incorrect results if they are counted more than once.

```python
# Check the number of duplicate rows
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

### **Observation**

### The duplicate-record check has been completed. Duplicate rows identified in the raw dataset will be removed so that each application record is counted only once during analysis.

# **Removing Duplicate Records**

### **Purpose**

### Remove duplicate records from the dataset to ensure that each application entry is represented only once in the analysis.

```python
# Remove duplicate rows
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

### **Observation**

### The duplicate records have been removed from the dataset. This prevents the same application record from being counted multiple times during the analysis.

# **Checking Invalid Rating Values**

### **Purpose**

### Identify rating values outside the valid 1–5 range.

### Invalid rating values can affect statistical analysis and visualizations, so they need to be identified and handled before further analysis.

```python
# Check for ratings outside the valid range
invalid_ratings = df[(df["Rating"] < 1) | (df["Rating"] > 5)]

print("Number of invalid rating records:", len(invalid_ratings))
print(invalid_ratings[["App", "Rating"]])
```

### **Observation**

### One invalid rating value was identified in the dataset. The rating of 19.0 is outside the valid range of 1–5, so this record should be removed before further analysis.

# **Removing Invalid Rating Records**

### **Purpose**

### Remove records containing rating values outside the valid 1–5 range to ensure that the Rating column contains only valid values for further analysis.

```python
# Remove records with invalid ratings
df = df[(df["Rating"] >= 1) & (df["Rating"] <= 5)]

print("Shape after removing invalid ratings:", df.shape)
```

### **Observation**

### The invalid rating record has been removed. The Rating column now contains only values within the valid 1–5 range.

# Converting Reviews to Numeric

### **Purpose**

### Convert the Reviews column from text format to numeric format so that it can be used for statistical analysis, calculations, and visualizations.

```python
# Convert Reviews from text to numeric
df["Reviews"] = pd.to_numeric(df["Reviews"], errors="coerce")

print("Reviews data type:", df["Reviews"].dtype)
print("Missing Reviews after conversion:", df["Reviews"].isnull().sum())
```

### **Observation**

### The Reviews column has been successfully converted from text to numeric format. No missing values were introduced during the conversion, so the column is ready for numerical analysis.




# Converting Installs to Numeric

### **Purpose**

### Clean the Installs column by removing symbols and separators, then convert it from text format to numeric format for analysis and calculations.

```python
# Remove "+" and "," from Installs
df["Installs"] = df["Installs"].str.replace("+", "", regex=False)
df["Installs"] = df["Installs"].str.replace(",", "", regex=False)

# Convert Installs to numeric
df["Installs"] = pd.to_numeric(df["Installs"], errors="coerce")

print("Installs data type:", df["Installs"].dtype)
print("Missing Installs after conversion:", df["Installs"].isnull().sum())
```

### **Observation**

### The Installs column has been cleaned by removing the + and comma separators and converted into a numeric format. This allows installation counts to be used correctly in calculations, statistical analysis, and visualizations.

# Converting Price to Numeric

### **Purpose**

### Clean the Price column by removing the dollar symbol and convert it from text format to numeric format for pricing analysis.

```python
# Remove "$" from Price
df["Price"] = df["Price"].str.replace("$", "", regex=False)

# Convert Price to numeric
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")

print("Price data type:", df["Price"].dtype)
print("Missing Price after conversion:", df["Price"].isnull().sum())
```

### **Observation**

### The Price column has been successfully cleaned by removing the $ symbol and converting the values into numeric format. The column is now ready for pricing analysis.

# Handling Application Size

### Purpose

Convert the Size column into a numeric format while separately identifying applications whose size varies with the device.

The Size_Varies indicator preserves this information before the text values are converted to missing values.

```python
# Identify apps whose size varies with the device
df["Size_Varies"] = (df["Size"] == "Varies with device").astype(int)

print("Apps with variable size:", df["Size_Varies"].sum())
```

### **Observation**

### The Size_Varies indicator has been created to identify apps whose size is listed as “Varies with device.” This preserves that information before converting the Size column into a numeric format.

# Converting Size to Numeric

### **Purpose**

### Convert the Size column into numeric values by removing the M unit.

### Values recorded as "Varies with device" will become missing values during conversion and will be handled separately using an appropriate imputation method.

```python
# Remove the "M" unit
df["Size"] = df["Size"].str.replace("M", "", regex=False)

# Convert Size to numeric
df["Size"] = pd.to_numeric(df["Size"], errors="coerce")

print("Size data type:", df["Size"].dtype)
print("Missing Size values after conversion:", df["Size"].isnull().sum())
```

# Imputing Missing Size Values

### **Purpose**

## Handle the missing values created during the conversion of the Size column.

### The median application size is used for imputation because it is less affected by unusually large or small app sizes.

```python
# Calculate the median application size
size_median = df["Size"].median()

# Fill missing Size values with the median
df["Size"] = df["Size"].fillna(size_median)

print("Median Size:", size_median)
print("Missing Size values after imputation:", df["Size"].isnull().sum())
```

### **Observation**

### The Size column has now been converted to numeric values, and the missing values created from "Varies with device" have been replaced with the median application size.

# Converting Last Updated to Date Format

### **Purpose**

### Convert the Last Updated column from text format to datetime format so that update dates can be analyzed and features such as update year can be created.

```python
# Convert Last Updated to datetime
df["Last Updated"] = pd.to_datetime(df["Last Updated"], errors="coerce")

print("Last Updated data type:", df["Last Updated"].dtype)
print("Missing Last Updated after conversion:", df["Last Updated"].isnull().sum())
```

### **Observation**

### The Last Updated column has been successfully converted from text format to datetime format. No missing values were introduced during the conversion, so it is ready for date-based analysis.

# **FEATURE ENGINEERING**

# Feature Engineering

### **Reviews per Install**

### **Purpose**

### Create a new feature called Reviews_per_Install to measure user engagement relative to the number of app installations.

### A higher value indicates that a larger proportion of installed users have submitted reviews.

```python
# Calculate Reviews per Install
df["Reviews_per_Install"] = np.where(
    df["Installs"] > 0,
    df["Reviews"] / df["Installs"],
    0
)

print(df[["App", "Reviews", "Installs", "Reviews_per_Install"]].head())
```

### **Observation**

### The Reviews_per_Install feature has been created successfully. It measures the number of reviews relative to app installations and will help us study user engagement during the EDA.

# Feature Engineering

### **Estimated Revenue**

### **Purpose**

### Create an Estimated_Revenue feature to estimate potential gross revenue from app installations and listed price.

### This metric provides a simple way to compare the potential revenue associated with free and paid applications.

```python
# Calculate estimated revenue
df["Estimated_Revenue"] = df["Installs"] * df["Price"]

print(df[["App", "Type", "Installs", "Price", "Estimated_Revenue"]].head())
```

### **Observation**

### The Estimated_Revenue feature has been created successfully by multiplying Installs by Price. Free applications have zero estimated revenue, while paid applications will have estimated revenue based on their listed price and installation count.


# Feature Engineering

### **Update Year**

### **Purpose**

### Create a new Updated_Year feature by extracting the year from the Last Updated date.

### This feature will be used to study app update trends over time.

```python
# Extract the year from Last Updated
df["Updated_Year"] = df["Last Updated"].dt.year

print(df[["App", "Last Updated", "Updated_Year"]].head())
```

### **Observation**

### The Updated_Year feature has been successfully created by extracting the year from Last Updated. This will allow us to analyze app update activity and trends over the years.
